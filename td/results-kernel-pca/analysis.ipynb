```python
import pandas as pd

# 1. Wczytanie danych z pliku CSV
df = pd.read_csv("pca_100_components_lifelines.csv")

# 2. Znalezienie indeksu wiersza z najwyższym wynikiem na zbiorze testowym
best_idx = df['ci_testing'].idxmax()

# 3. Wyciągnięcie odpowiednich parametrów
best_ci_test = df.loc[best_idx, 'ci_testing']
best_penalizer = df.loc[best_idx, 'penalizer']
best_l1_ratio = df.loc[best_idx, 'l1_ratio'] # Zwróć uwagę na l1_ratio!
best_ci_train = df.loc[best_idx, 'ci_training']

print("--- NAJLEPSZY MODEL ---")
print(f"Najwyższe ci_testing: {best_ci_test}")
print(f"Penalizer: {best_penalizer}")
print(f"L1 Ratio:  {best_l1_ratio}")
print(f"(ci_training dla tego modelu: {best_ci_train})")
```
