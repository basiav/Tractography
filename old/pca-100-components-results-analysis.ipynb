## Cosine Kernel PCA

```python
import pandas as pd


df_lifelines = pd.read_csv("results/pca_100_components_lifelines.csv")
df_sksurv_ridge = pd.read_csv("results/kernel_pca_cosine_100_comps_scikit_surv_ridge.csv")
df_sksurv_elastic = pd.read_csv("results/kernel-pca-100-comps-scikit-surv-elastic.csv")
df_sksurv_rf = pd.read_csv("results/kernel-pca-100-comps-scikit-random-forest.csv")


best_combinations_lifelines = df_lifelines.loc[df_lifelines.groupby('penalizer')['ci_testing'].idxmax()]
best_combinations_sksurv_ridge = df_sksurv_ridge.loc[df_sksurv_ridge.groupby('alpha')['ci_testing'].idxmax()]
best_combinations_skurv_elastic = df_sksurv_elastic.loc[df_sksurv_elastic.groupby(['alpha', 'l1_ratio'])['ci_testing'].idxmax()]
best_combinations_pca_skurv_survival_forest = df_sksurv_rf.loc[df_sksurv_rf.groupby(['n_estimators', 'min_samples_leaf', 'min_samples_split'])['ci_testing'].idxmax()]


best_combinations_lifelines = best_combinations_lifelines.sort_values(by='ci_testing', ascending=False)
best_combinations_sksurv_ridge = best_combinations_sksurv_ridge.sort_values(by='ci_testing', ascending=False)
best_combinations_skurv_elastic = best_combinations_skurv_elastic.sort_values(by='ci_testing', ascending=False)
best_combinations_pca_skurv_survival_forest = best_combinations_pca_skurv_survival_forest.sort_values(by='ci_testing', ascending=False)


display(best_combinations_lifelines[['penalizer', 'l1_ratio', 'ci_training', 'ci_testing']].head(1))
display(best_combinations_sksurv_ridge[['alpha',  'ci_training', 'ci_testing']].head(1))
display(best_combinations_skurv_elastic[['alpha', 'l1_ratio', 'ci_training', 'ci_testing']].head(1))
display(best_combinations_pca_skurv_survival_forest[['n_estimators', 'min_samples_leaf', 'min_samples_split', 'ci_training', 'ci_testing']].head(1))
```
